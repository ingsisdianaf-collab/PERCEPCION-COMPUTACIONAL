```python
# Funciones de apoyo para visualizar imágenes

def mostrar(imagen, titulo="Imagen", cmap=None):
    """Muestra una imagen OpenCV en Colab/Matplotlib."""
    plt.figure(figsize=(7, 5))
    if imagen.ndim == 2:
        plt.imshow(imagen, cmap=cmap or "gray")
    else:
        plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
    plt.title(titulo)
    plt.axis("off")
    plt.show()
```

```python
# Tratamiento de imagen problemática
# Filtro Laplaciano (realce de bordes)

imagen = cv2.imread('desenfocada.jpeg')

# Convertir a escala de grises
gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

# Aplicar filtro Laplaciano para realzar bordes
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
sharp = cv2.convertScaleAbs(gray - laplacian)

# Mostrar resultados
plt.figure(figsize=(10,5))
plt.subplot(1,3,1); plt.imshow(gray, cmap="gray"); plt.title("desenfocada")
plt.subplot(1,3,2); plt.imshow(laplacian, cmap="gray"); plt.title("Laplaciano")
plt.subplot(1,3,3); plt.imshow(sharp, cmap="gray"); plt.title("Imagen mejorada")
plt.show()
```
